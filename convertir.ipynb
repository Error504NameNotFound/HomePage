```python
# 1. Instalar la librería requerida
!pip install -q pymupdf4llm

import pathlib
import pymupdf4llm
from google.colab import files

# 2. Subir tus archivos PDF desde la computadora
print("Selecciona y sube todos tus archivos PDF:")
uploaded = files.upload()

# 3. Convertir cada PDF subido a un archivo Markdown (.md)
pdf_files = sorted(list(pathlib.Path(".").glob("*.pdf")))

print(f"\n--- Procesando {len(pdf_files)} archivos PDF ---\n")

for pdf_file in pdf_files:
    try:
        # Convertir el PDF a texto en formato Markdown
        md_text = pymupdf4llm.to_markdown(str(pdf_file))

        # Definir nombre del archivo de salida .md
        output_file = pdf_file.with_suffix(".md")

        # Guardar el resultado
        with open(output_file, "w", encoding="utf-8") as f:
            f.write(md_text)

        print(f"✔ Convertido: {output_file.name}")

        # Descargar el archivo .md a tu computadora automáticamente
        files.download(str(output_file))

    except Exception as e:
        print(f"✖ Error en {pdf_file.name}: {e}")

print("\n¡Proceso finalizado con éxito!")
```
